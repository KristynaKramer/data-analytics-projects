#### Filling missing data

```python
df_clean[df_clean['Entity'] == 'Canada'].head(25)
```

We want to fill-in missing years and deal with nan values. The most reasonable way is to fill the data from the last known year.

```python
# Each entity has one fixed World region
entity_region = df_clean.drop_duplicates('Entity')[['Entity', 'Code', 'World region']]

all_entities = df_clean['Entity'].unique()
last_year = df_clean['Year'].max()
all_years = range(START_YEAR, last_year + 1)

# Full Entity x Year grid
full_index = pd.MultiIndex.from_product([all_entities, all_years], names=['Entity', 'Year'])
df_grid = pd.DataFrame(index=full_index).reset_index()

# Merge in the real data (missing combos become NaN)
df_full = df_grid.merge(
    df_clean.drop(columns=['World region', 'Code']),  # drop to avoid duplicate col, re-add below
    on=['Entity', 'Year'],
    how='left'
)

# Re-attach the fixed World region per entity (so it's never NaN, even for missing years)
df_full = df_full.merge(entity_region, on='Entity', how='left')


df_full = df_full.sort_values(['Entity', 'Year'])
columns = df_full.columns
df_filled = df_full.groupby('Code')[columns].ffill()
```

## Explore data by continent

We don't have the continental data for North and South America (only for 'Americas'). We need to compute these data. And, to be consistent, we will compute them for all continents.

- Population = Sum(Population)
- GDP per capita = Sum(Population * GDP per capita) / Sum(Population)
- Life expectancy = Sum(Life expectancy * Population) / Sum(Population)


```python
years = df_clean['Year'].unique()
rows = []

for region in regions:
    for year in years:
        df_cy = df_clean[(df_clean['World region'] == region) & (df_clean['Year'] == year)]

        population = df_cy['Population'].sum()
        population_gdp = df_cy.loc[df_cy['GDP per capita'].notna(), 'Population'].sum()
        population_life = df_cy.loc[df_cy['Life expectancy at birth'].notna(), 'Population'].sum()

        if population_gdp > 0:
            gdp_per_capita = (df_cy['GDP per capita'] * df_cy['Population']).sum() / population_gdp
        else:
            gdp_per_capita = np.nan

        if population_life > 0:
            life_expectancy = (df_cy['Life expectancy at birth'] * df_cy['Population']).sum() / population_life
        else:
            life_expectancy = np.nan

        if gdp_per_capita == 0:
            gdp_per_capita = np.nan

        if life_expectancy == 0:
            life_expectancy = np.nan
        
        rows.append({
            'World region': region,
            'Year': year,
            'Population': population,
            'GDP per capita': gdp_per_capita,
            'Life expectancy at birth': life_expectancy
        })

df_regions = pd.DataFrame(rows)
df_regions.info()
```

### Life expectancy by continent -- histograms
We want to figure out common years for all regions.

Thanks to filling-in the data, we know that once we have a notna datapoint for a region, we have continuous data for that region. Hence, it is sufficient to find the region with the latest start.

```python
regions_starts = []

for region in regions:
    regions_starts.append(
        df_regions_life[df_regions_life['World region'] == region]['Year'].min()
    )

regions_start_year = max(regions_starts)
regions_end_year = df_regions_life['Year'].max()
print(regions_start_year, regions_end_year)
```

```python
# The data are only consistent from 1950
regions_start_year = 1950

positions = {
    'Asia': (1, 1),
    'Europe': (1, 2),
    'Africa': (1, 3),
    'Oceania': (2, 1),
    'North America': (2, 2),
    'South America': (2, 3),
}

xbins = dict(start=30, end=85, size=3)


# Build the initial figure using the first year

year = regions_start_year
df_year = df_clean[df_clean['Year']==year]

fig = subplots.make_subplots(rows=2, cols=3, subplot_titles=regions, shared_yaxes=True)

for region, (row,col) in positions.items():
    trace = go.Histogram(
                    x = df_year[df_year['World region'] == region]['Life expectancy at birth'],
                    y=df_year[df_year['World region'] == region]['Population'],
                    histfunc='sum',
                    histnorm='percent',
                    xbins=xbins,
                    marker_color = colors_hex[region],
                    name=''
    )
    fig.add_trace(trace=trace, row=row, col=col)

# Build one frame per year — trace order must match the order added above
frames = []
years = range(regions_start_year, regions_end_year + 1)
for year in years:
    df_year = df_clean[df_clean['Year'] == year]
    frame_data = [
            go.Histogram(
                x=df_year[df_year['World region'] == region]['Life expectancy at birth'],
                y=df_year[df_year['World region'] == region]['Population'],
                histfunc='sum',
                histnorm='percent',
                xbins=xbins,
                marker_color=colors_hex[region],
                name=''
            )
            for region in positions.keys()
    ]
    frames.append(go.Frame(data=frame_data, name=str(year)))

fig.frames = frames

# Slider
sliders = [dict(
    active=0,
    steps=[
        dict(
            method="animate",
            args=[[str(year)], dict(mode="immediate",
                                     frame=dict(duration=300, redraw=True),
                                     transition=dict(duration=0))],
            label=str(year)
        )
        for year in years
    ],
    x=0.1, y=-0.05, len=0.9
)]

buttons = [
        dict(
            label="▶",
            method="animate",
            args=[None, dict(frame=dict(duration=300, redraw=True),
                                fromcurrent=True,
                                transition=dict(duration=0))]
        ),
        dict(
            label="■",
            method="animate",
            args=[[None], dict(frame=dict(duration=0, redraw=False),
                                mode="immediate",
                                transition=dict(duration=0))]
        ),
    ]

menu = [dict(
        type="buttons",
        direction="left",
        x=0.1, y=-0.15,
        showactive=False,
        buttons=buttons
    )]



title = '<b>Development of life expectancy at birth by world region</b>'
fig.update_layout(
    title=title, 
    title_x=0.5, 
    showlegend=False, 
    sliders=sliders, 
    updatemenus=menu,
    width=WIDTH,
    height=HEIGHT,
    margin=dict(l=80, r=20, t=70, b=20)
    )
fig.update_xaxes(range=[30, 85])
fig.update_xaxes(title_text = 'Life expectancy (years)', row=2, col=2)
fig.update_yaxes(range=[0, 100], ticksuffix='%')
#fig.update_yaxes(title_text = 'Of total population', row=1, col=1)
fig.add_annotation(
    text="Share of total population of the world region",
    xref="paper", yref="paper",
    x=-0.11, y=0.5,
    showarrow=False,
    textangle=-90,
    font=dict(size=14),
)

pyo.plot(fig, filename='histograms_life.html', config=dict(displayModeBar=True), auto_open=False, auto_play=False)
fig.show()
```
